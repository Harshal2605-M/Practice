```python
employees = [
    {"name": " Alice ", "salary": "75000", "department": "IT"},
    {"name": "Bob", "salary": "55000", "department": " HR "},
    {"name": "", "salary": "60000", "department": "Finance"},
    {"name": "David", "salary": "invalid", "department": "IT"},
    {"name": "Eva", "salary": "85000", "department": ""},
    {"name": " Frank ", "salary": "70000", "department": "HR"}
]
```

```python
def clean_employees(employees):
    cleaned = []

    for employee in employees:

        # strip name and department
        employee["name"] = employee["name"].strip()
        employee["department"] = employee["department"].strip()

        # check empty name/department
        if (employee["name"] == "" ) or (employee["department"] == ""): 
            continue
        # try to convert salary
        try :
            employee["salary"] = float(employee["salary"])

        except:
            continue
                

        # if everything is valid:
        #     append employee
        cleaned.append(employee)

    return cleaned
```

```python
clean_employees(employees)
```
