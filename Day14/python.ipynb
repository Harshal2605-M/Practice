```python
"""Practical Question 1

Create two classes:

DataScientist

Method:

work()

Should print:

Data Scientist → Building ML models
DataEngineer

Method:

work()

Should print:

Data Engineer → Building data pipelines

Create objects:

ds = DataScientist()
de = DataEngineer()

Then put both objects into:

employees = [ds, de]

and use one for loop to call work()."""
```

```python
class DataScientist :
    def __int__(self):
        pass

    def work(self):
        print("Data Scientist → Building ML models")

class DataEngineer :
    def __int__(self):
            pass
    
    def work(self):
        print("Data Engineer → Building data pipelines")

ds = DataScientist()
de = DataEngineer()
employees = [ds, de]
for employee in employees :
     employee.work()
```
