```python
employees = [
    {"name": "Alice", "salary": 90000, "department": "IT"},
    {"name": "Bob", "salary": 65000, "department": "HR"},
    {"name": "Charlie", "salary": 80000, "department": "IT"},
    {"name": "David", "salary": 55000, "department": "Finance"}
]
```

```python
"""Task: Return the names of employees whose:

department is "IT"
salary is greater than 70000

Use one filter() and one map()."""
```

```python
employees = list(filter(lambda employee: employee["department"] == "IT" and employee["salary"] > 70000, employees))
employees = list(map(lambda employee: employee["name"] , employees))
print(employees)
```
