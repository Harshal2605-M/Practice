```python
"""Given:

employees = [
    " Alice | Python,SQL,Pandas | 100000 ",
    " Bob | Python,NumPy | 80000 ",
    "CHARLIE | SQL,Pandas,Power BI | 70000 ",
    " David | Python,SQL,Scikit-learn | -5000 "
]

Write a program that:

Cleans the spaces.
Converts names to proper case.
Converts skills into a list.
Converts salary to int.
If salary is negative, raise a ValueError.
Handle the exception without stopping the entire loop.
For valid employees, print:
Alice -> ['Python', 'SQL', 'Pandas'] -> 100000
Bob -> ['Python', 'NumPy'] -> 80000
Charlie -> ['SQL', 'Pandas', 'Power BI'] -> 70000"""
```

```python
employees = [
    " Alice | Python,SQL,Pandas | 100000 ",
    " Bob | Python,NumPy | 80000 ",
    "CHARLIE | SQL,Pandas,Power BI | 70000 ",
    " David | Python,SQL,Scikit-learn | -5000 "
]

for employee in employees :
    try :
        name, skills, salary = employee.split("|")
        name = name.strip().title()
        skills = skills.strip().split(',')
        salary = int(salary.strip())
        if salary < 0 :
            raise ValueError("Salary cannot be negative.")
        print(f"{name} -> {skills} -> {salary}")
    except ValueError as e:
        print(f"Error: {e}")

```
