```python
"""Q1 — Try yourself

Given:

numbers = [1, 2, 3, 4, 5, 6]

Create a list containing the squares of only the even numbers using list comprehension."""
```

```python
numbers = [1, 2, 3, 4, 5, 6]
square = [a*a for a in numbers]
print(square)
```

```python
odd = [a*10 for a in numbers if a % 2 != 0]
print(odd)
```

```python
numbers = [1, 2, 3, 4, 5]
square = {a : a*a for a in numbers}
```

```python
employees = [
    {"name": "Alice", "salary": 100000},
    {"name": "Bob", "salary": 80000},
    {"name": "Charlie", "salary": 70000}
]
employees_dict = {employee['name']: employee['salary'] for employee in employees}
print(employees_dict)
```

```python
employees_dict_gt_80000 = {employee['name']: employee['salary'] for employee in employees if employee['salary'] >= 80000}
print(employees_dict_gt_80000)
```

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]
even = [a for row in matrix for a in row if a % 2 == 0]
print(even)
```
