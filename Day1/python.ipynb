```python
"""Scenario: Process Student Dataset

You're given the following data:

students = [
    {"id": 101, "name": "Alice", "marks": [85, 92, 78]},
    {"id": 102, "name": "Bob", "marks": [65, 70, 60]},
    {"id": 103, "name": "Charlie", "marks": [95, 90, 93]},
    {"id": 104, "name": "David", "marks": [40, 55, 50]},
    {"id": 105, "name": "Eva", "marks": [88, 76, 91]}
]
Tasks

Write reusable, clean Python code (don't hardcode values).

Calculate the average marks of every student.
Add a new key called "average" to each dictionary.
Add another key "result":
"Pass" if average ≥ 60
"Fail" otherwise
Return a new list sorted by average marks in descending order.
Print only the names of students who scored more than 90 in at least one subject.
Rules
✅ Use functions.
✅ Don't modify the original list directly.
✅ Write readable code.
❌ Don't use Pandas."""
```

```python
students = [
    {"id": 101, "name": "Alice", "marks": [85, 92, 78]},
    {"id": 102, "name": "Bob", "marks": [65, 70, 60]},
    {"id": 103, "name": "Charlie", "marks": [95, 90, 93]},
    {"id": 104, "name": "David", "marks": [40, 55, 50]},
    {"id": 105, "name": "Eva", "marks": [88, 76, 91]}
]
```

```python
def avg(marks):
    return sum(marks)/ len(marks)
```

```python
def result(marks):
    if marks > 60 :
        return f"Pass"
    else :
        return f"Fail"
```

```python
new_students = []
```

```python
for student in students:
    new_student = student.copy()
    average = avg(student["marks"])
    new_student["average"] = average
    new_student["result"] = result(average)
    new_students.append(new_student)
```

```python
print(new_students)
```

```python
print(students)
```

```python
new_students = sorted(
    new_students,
    key=lambda student: student["average"],
    reverse=True
)
```

```python
print(new_students)
```

```python
for student in new_students:
    if any(mark > 90 for mark in student["marks"]):
        print(student["name"])
```
