```python
"""Question 1: map() + lambda

Given:

salaries = [45000, 55000, 65000, 75000, 85000]

Create a new list where every salary is increased by 10%.

Expected:

[49500.0, 60500.0, 71500.0, 82500.0, 93500.0]"""
```

```python
salaries = [45000, 55000, 65000, 75000, 85000]
salaries = list(map(lambda a: a + a*.10, salaries))
print(salaries)
```

```python
salaries = [45000, 55000, 65000, 75000, 85000]
salaries = list(filter(lambda a: a > 60000, salaries))
print(salaries)
```

```python
salaries = [45000, 55000, 65000, 75000, 85000]
salaries = list(filter(lambda a: a>60000, salaries))
salaries = list(map(lambda a: a + a*0.10, salaries))
print(salaries)
```

```python
employees = list(filter(lambda employee: employee["salary"] > 70000, employees))
employees_names = list(map(lambda employee: employee["name"], employees))
print(employees_names)
```

```python
employees_it = list(filter(lambda employee:employee["department"] == "IT", employees))
employees = list(filter(lambda employee: employee["salary"] > 70000, employees_it))
employees_names = list(map(lambda employee: employee["name"], employees))
print(employees_names)
```

```python
employees = [
    {"name": "Alice", "salary": 75000, "department": "IT"},
    {"name": "Bob", "salary": 55000, "department": "HR"},
    {"name": "Charlie", "salary": 90000, "department": "IT"},
    {"name": "David", "salary": 65000, "department": "Finance"},
    {"name": "Eva", "salary": 85000, "department": "IT"}
]
```

```python
employees = list(map(lambda employee: employee["salary"]*1.10, employees_it))
print(employees)
```

```python
employees_updated = list(
    map(
        lambda employee: {
            **employee,
            "salary": employee["salary"] * 1.10
        },
        employees_it
    )
)
print(employees_updated)
```
