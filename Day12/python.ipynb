```python
"""Q1 — Employee Class

Create an Employee class with:

name
salary
department

Use __init__().

Create 3 employees:

Alice → 100000 → IT
Bob → 80000 → HR
Charlie → 90000 → Finance

Then print each employee's:

name
salary
department
Q2 — Add a method

Add:

display_info()

Instead of separately writing:

print(emp1.name)
print(emp1.salary)

you should be able to do:

emp1.display_info()"""
```

```python
class Employee :
    def __init__(self,name,salary,department) :
        self.name = name 
        self.salary = salary
        self.department = department

    def display_info(self) :
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Department:", self.department)

emp1 = Employee("Harshal", 60000, "IT")
emp1.display_info()
emp2 = Employee("Aditya", 70000, "HR")
emp2.display_info()
emp3 = Employee("Yash", 50000, "IT")
emp3.display_info()
```

```python
"""Q3 — Add a calculation method

Add:

annual_salary()

which returns:

salary × 12

For example:

emp1.annual_salary()

should return:

1200000"""
```

```python
class Employee:

    def __init__(self, name, salary, department):
        self.name = name
        self.salary = salary
        self.department = department

    def display_info(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Department:", self.department)

    def annual_income(self):
        return self.salary * 12


emp1 = Employee("Harshal", 60000, "IT")

emp1.display_info()

print("Annual Income:", emp1.annual_income())
```

```python
"""Q4 — Practical ML/Data Engineer scenario

Create a DataScientist class with:

name
salary
skills

Example:

ds1 = DataScientist(
    "Alice",
    120000,
    ["Python", "SQL", "Pandas", "Scikit-learn"]
)

Create a method:

skill_count()

that returns the number of skills.

Expected:

4"""
```

```python
class DataScientist:

    def __init__(self, name, salary, skills):
        self.name = name
        self.salary = salary
        self.skills = skills

    def display_info(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Skills:", self.skills)

    def skill_count(self):
        return len(self.skills)


ds1 = DataScientist(
    "Alice",
    120000,
    ["Python", "SQL", "Pandas", "Scikit-learn"]
)

ds1.display_info()

print("Skills Count:", ds1.skill_count())
```

```python
"""Create a MachineLearningModel class with:

model_name
algorithm
accuracy

Methods:

display_info() → prints all details
is_good_model() → returns True if accuracy is >= 0.80, otherwise False

Create two objects:

Random Forest → accuracy 0.92
Logistic Regression → accuracy 0.74

Then print whether each model is good."""
```

```python
class MachineLearningModel :
    def __init__(self,algorithm,accuracy):
        self.algorithm = algorithm
        self.accuracy = accuracy

    def display_info(self):
        print("Algorithm:", self.algorithm)
        print("Accuracy:", self.accuracy)

    def is_good_model(self):
        if self.accuracy >= 0.80 :
             return "good"
        else :
            return "bad" 

m1 = MachineLearningModel("Random Forest", 0.92)
m2 = MachineLearningModel("Logistic Regression", 0.74)
m1.display_info()
print("Model Good or Bad:", m1.is_good_model())
m2.display_info()
print("Model Good or Bad:", m2.is_good_model())
```
