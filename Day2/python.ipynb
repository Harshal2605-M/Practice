```python
"""Scenario

You receive a CSV file containing employee data:

id,name,department,salary
101,Alice,IT,75000
102,Bob,HR,55000
103,Charlie,IT,90000
104,David,Finance,65000
105,Eva,IT,85000
106,Frank,HR,60000
Your Task

Using only Python — do not use Pandas:

Read the CSV file.
Calculate the average salary.
Find employees whose salary is greater than the average salary.
Group employees by department.
Calculate the average salary of each department.

Write the employees earning above the overall average into a new CSV file called:

high_salary_employees.csv"""
```

```python
import csv
```

```python
def read_employees(filename):
    employees_list = []
    with open(filename, "r") as f :
        reader = csv.DictReader(f)

        for row in reader:
            employees_list.append(row)

    return employees_list
```

```python
employees = read_employees("employees.csv")

print(employees)
print(len(employees))
```

```python
def avg_salary(employees) :
    total_salary  = 0 
    for employee in employees :
        salary = float(employee["salary"])
        total_salary  += salary

    return total_salary /len(employees)
```

```python
avg_salary(employees)
```

```python
def high_salary_employees(employees, average):
    result = []
    for employee in employees :
        salary = float(employee["salary"])
        if(salary > average) :
            result.append(employee["name"])
    return result
```

```python
high_salary_employees(employees, avg_salary(employees))
```

```python
def grp_by_dept(employees) :
    department_salaries = {}
    for employee in employees :
        department = employee["department"]
        salary = float(employee["salary"])

        if department not in department_salaries:
            department_salaries[department] = []

        department_salaries[department].append(salary)

    return department_salaries
```

```python
grp_by_dept(employees)
```

```python
def avg_salary_dept(salaries):
    return sum(salaries)/len(salaries)
```

```python
def avg_salary_by_dept(departments) :
    department_salaries = {}
    for dept, salaries in departments.items() :
        average = avg_salary_dept(salaries)
        department_salaries[dept] = average

    return department_salaries

avg_salary_by_dept(grp_by_dept(employees))
```

```python
def high_salary_employees_csv(employees, average):
    result = []
    for employee in employees :
        salary = float(employee["salary"])
        if(salary > average) :
            result.append(employee)
    return result

high_salary = high_salary_employees_csv(employees, avg_salary(employees))
```

```python
with open("high_salary_employees.csv", "w", newline="") as f :
    fieldnames = ["id", "name", "department", "salary"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    for employee in high_salary:
        writer.writerow(employee)
```
