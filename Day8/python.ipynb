```python
def calculate_average(*args) :
    return sum(args) / len(args)

print(calculate_average(10, 20, 30))
print(calculate_average(10, 20, 30, 40, 50))
```

```python
"""🔥 Final Python Functions Question

This is closer to an actual interview coding question.

def calculate_total(*prices, discount=0):
    # write your code

Requirements:

Accept any number of prices.
Calculate the total.
Apply the discount percentage.
Return the final amount."""
```

```python
def calculate_total(*prices, discount=0):
    total = sum(prices)
    discount_amount = total * (discount/100)
    final = total - discount_amount
    print("Total :",total)
    print("Discount :", discount_amount)
    print("Final Amount :", final)

calculate_total(100, 200, 300, discount=10)
```
