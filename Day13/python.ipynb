```python
"""Parent class: Employee

Attributes:

name
salary

Method:

display_info()
Child class: DataScientist

Additional attribute:

skills

Additional method:

skill_count()

Create:

ds1 = DataScientist(
    "Alice",
    120000,
    ["Python", "SQL", "Pandas", "Scikit-learn"]
)

Then call:

ds1.display_info()
print(ds1.skill_count())"""
```

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_info(self):
        print("Name:", self.name)
        print("Salary:", self.salary)


class DataScientist(Employee):

    def __init__(self, name, salary, skills):
        super().__init__(name, salary)
        self.skills = skills

    def skill_count(self):
        return len(self.skills)


ds1 = DataScientist(
    "Alice",
    120000,
    ["Python", "SQL", "Pandas", "Scikit-learn"]
)

ds1.display_info()
print("Skill Count:", ds1.skill_count())
```

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_role(self):
        print("Employee: General Employee")
        


class DataScientist(Employee):
    def display_role(self):
            print("Employee: Data Scientist")
    


emp = Employee("Bob", 80000)
ds = DataScientist("Alice", 120000)

emp.display_role()
ds.display_role()
```
