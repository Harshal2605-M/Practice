```python
employees = [
    {"name": "Alice", "department": "IT", "salary": 75000},
    {"name": "Bob", "department": "HR", "salary": 55000},
    {"name": "Charlie", "department": "IT", "salary": 90000},
    {"name": "David", "department": "Finance", "salary": 65000},
    {"name": "Eva", "department": "IT", "salary": 85000},
    {"name": "Frank", "department": "HR", "salary": 60000}
]
```

```python
def total_salary(salaries):
    return sum(salaries)
```

```python
def avg_salary(salaries):
    return sum(salaries)/len(salaries)
```

```python
def count_employee(salaries):
    return len(salaries)
```

```python
def grp_by_dept(employees):
    department_salaries = {}
    for employee in employees:
        department = employee["department"] 
        salary = employee["salary"]

        if department not in department_salaries :
            department_salaries[department] = []

        department_salaries[department].append(salary)

    return department_salaries
```

```python
grp_by_dept(employees)
```

```python
def department_summary(employees):
    department_summary = {}

    department_salaries = grp_by_dept(employees)
    for dept, salaries in department_salaries.items() :
        depart = {}
        count = count_employee(salaries)
        total_sal = total_salary(salaries)
        avg_sal = avg_salary(salaries)


        depart["count"] = count
        depart["total_salary"] = total_sal
        depart["average_salary"] = avg_sal

        department_summary[dept] = depart

    return department_summary
```

```python
department_summary(employees)
```
