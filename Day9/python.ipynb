```python
"""🔥 Practical Scenario — Employee Skills Data

You receive employee data as strings from an API:

employees = [
    "Alice | Python,SQL,Pandas",
    " Bob | Python,NumPy,Scikit-learn ",
    "CHARLIE | SQL,Pandas,Power BI"
]

You need to clean and process this data.

Your task

For each employee, produce:

Remove unnecessary spaces.
Convert the employee's name to proper case.
Extract the skills.
Convert the skills string into a list.
Count the number of skills.
Create a final output like:"""
```

```python
employees = [
    "Alice | Python,SQL,Pandas",
    " Bob | Python,NumPy,Scikit-learn ",
    "CHARLIE | SQL,Pandas,Power BI"
]
for employee in employees :
    name, skills= employee.split("|")
    name = name.strip().title()
    skills = skills.strip()
    skills = skills.split(",")
    length = len(skills)
    print(f"{name} -> {skills} -> {length} skills")
```

```python
"""Let's keep it placement-focused.

Given:

a = 10
b = 0


result = a / b


print(result)

Question: Modify this code using try and except so that instead of the program crashing, it prints:

Cannot divide by zero

Write the code yourself."""
```

```python
"""Let's do one practical placement-style question.

Create a function:

def check_salary(salary):
    # your code

If salary is less than 0, raise an exception with:

Salary cannot be negative

Otherwise print:

Valid salary"""
```

```python
def check_salary(salary):
    if salary < 0:
        raise ValueError("Salary cannot be negative")
    else:
        print("Valid salary")

check_salary(50000)

check_salary(-5000)
    
```
