```python
total_salary = 0 
with open("employees.txt", 'r') as f :
    while  True:
        content = f.readline()
        if not content:
                    break
        name, skill, salary = content.split(",")
        salary = int(salary)
        total_salary += salary

print(total_salary)
```

```python
with open("employees.txt", "r") as input_file:
    with open("high_salary.txt", "w") as output_file:

        while True:
            content = input_file.readline()

            if not content:
                break

            name, skill, salary = content.strip().split(",")
            salary = int(salary)

            if salary >= 80000:
                line = ",".join([name, skill, str(salary)])
                output_file.write(line + "\n")
```

```python
total_high_salary = 0

with open("employees.txt", "r") as input_file:
    with open("high_salary.txt", "w") as output_file:

        while True:
            content = input_file.readline()

            if not content:
                break

            name, skill, salary = content.strip().split(",")

            name = name.strip()
            skill = skill.strip()
            salary = int(salary.strip())

            if salary >= 80000:
                total_high_salary += salary

                line = ",".join([
                    name,
                    skill,
                    str(salary)
                ])

                output_file.write(line + "\n")

print("Total high salary:", total_high_salary)
```
